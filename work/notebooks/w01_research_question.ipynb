Search question:
Which content items show the strongest evidence of needing a refresh or further investigation, and can historical search-performance signals be used to prioritize a review queue?

Unit of analysis:
The primary unit of analysis will be an individual content item/page.

Decision:
Which pages should the SEO/content team review first?

Action:

Cost of a wrong recommendation:

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*



```python
import pandas as pd

url = "https://raw.githubusercontent.com/SaaDasim05/Flyrank-ML-Internship/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nColumns:")
print(df.columns.tolist())
```

```python
# Basic evidence for the Refresh / Content Opportunity lane

print(f"Total content items: {len(df):,}")

print(
    f"Average content age: "
    f"{df['content_age_days'].mean():.1f} days"
)

print(
    f"Average days since last update: "
    f"{df['days_since_last_update'].mean():.1f} days"
)

print("\nTrend direction:")
print(
    df["trend_direction"]
    .value_counts(normalize=True)
    .mul(100)
    .round(1)
)
```

### What I observed

The starter dataset contains **30,000 content items**. The average content age is **256.2 days**, while the average time since the last update is **46.1 days**.

The largest observed trend category is **down**, representing **54.2%** of the content items. This makes content performance changes and potential refresh opportunities worth investigating further.

These numbers do not show that older or recently updated content causes performance changes. They only show that the dataset contains substantial variation in content age, update recency, and observed search-performance direction, providing a useful starting point for the Refresh / Content Opportunity lane.
